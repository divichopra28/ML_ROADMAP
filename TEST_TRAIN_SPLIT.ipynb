**Performing Test,Train and split on Sample data**


 **Introduction**

Training Set → The data used to teach the machine learning model.
Testing Set → The data used to test how well the trained model performs on new, unseen data.

X → Input data (features) given to the model.
y → Output data (target) that the model needs to predict.
X_train → Input data used to train the model.
y_train → Correct output values used while training the model.
X_test → New input data used to test the trained model.
y_test → Actual output values used to evaluate the model's predictions.

**When we specify the test size , which is 0.25 , which means 25% of our data is for testing and the rest of the data is for Training**

```python
# import packages
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split

# importing data
df = pd.read_csv('headbrain1.csv')

# head of the data
print(df.head())

X= df['Head Size(cm^3)']
y=df['Brain Weight(grams)']

# using the train test split function
X_train, X_test,y_train, y_test = train_test_split(X,y ,random_state=104, test_size=0.25, shuffle=True)

# printing out train and test sets

print('X_train : ')
print(X_train.head())
print('')
print('X_test : ')
print(X_test.head())
print('')
print('y_train : ')
print(y_train.head())
print('')
print('y_test : ')
print(y_test.head()) 
```
